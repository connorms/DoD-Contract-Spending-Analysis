```python
import requests
```

```python
url = "https://api.usaspending.gov/api/v2/download/count/"
```

```python
payload = {
    "filters": {
        "time_period": [
            {
                "start_date": "2020-10-01",
                "end_date": "2021-09-30"
            }
        ],
        "recipient_scope": "domestic",
        "award_type_codes": ["A", "B", "C", "D"],
        "agencies": [
            {
                "type": "awarding",
                "tier": "toptier",
                "name": "Department of Defense"
            }
        ]
    }
}
```

```python
response = requests.post(url, json=payload)

print(response.status_code)

if response.status_code == 200:
    print(response.json())
else:
    print(response.text[:1000])
```

```python
prime_file = "Contracts_PrimeTransactions_2026-09-29_H04M07S08_1.csv"

with zipfile.ZipFile(zip_path, "r") as z:
    with z.open(prime_file) as f:
        prime_df = pd.read_csv(f)

print("Prime transaction rows:", len(prime_df))
print("Columns:", len(prime_df.columns))
```

```python
columns = [
    "contract_transaction_unique_key",
    "award_id_piid",
    "action_date",
    "federal_action_obligation",
    "total_dollars_obligated",
    "awarding_agency_name",
    "awarding_sub_agency_name",
    "awarding_office_name",
    "funding_agency_name",
    "funding_sub_agency_name",
    "recipient_name",
    "recipient_parent_name",
    "recipient_state_name",
    "primary_place_of_performance_state_name",
    "award_type",
    "type_of_contract_pricing",
    "naics_code",
    "naics_description",
    "product_or_service_code",
    "product_or_service_code_description",
    "extent_competed",
    "solicitation_procedures",
    "type_of_set_aside",
    "number_of_offers_received"
]
```

```python
analysis_df = prime_df[columns].copy()

print("Rows:", len(analysis_df))
print("Columns:", len(analysis_df.columns))
print(analysis_df.head())
```

```python
print(
    analysis_df[
        [
            "action_date",
            "federal_action_obligation",
            "recipient_name",
            "award_type",
            "naics_description"
        ]
    ].isna().sum()
)
```

```python
print(analysis_df["award_type"].value_counts())
```

```python
print(
    analysis_df.groupby("award_type")["federal_action_obligation"]
    .agg(["count", "sum", "mean", "median"])
    .sort_values("sum", ascending=False)
)
```

```python
contractor_summary = (
    analysis_df
    .groupby("recipient_name")["federal_action_obligation"]
    .agg(["count", "sum", "mean"])
    .sort_values("sum", ascending=False)
)

print(contractor_summary.head(15))
```

```python
parent_summary = (
    analysis_df
    .groupby("recipient_parent_name")["federal_action_obligation"]
    .agg(["count", "sum", "mean"])
    .sort_values("sum", ascending=False)
)

print(parent_summary.head(15))
```

```python
total_obligations = analysis_df["federal_action_obligation"].sum()

top_10_obligations = parent_summary.head(10)["sum"].sum()

top_10_share = top_10_obligations / total_obligations * 100

print(f"Total obligations: ${total_obligations:,.2f}")
print(f"Top 10 parent contractors: ${top_10_obligations:,.2f}")
print(f"Top 10 share: {top_10_share:.2f}%")
```

```python
for n in [1, 5, 10, 25, 50]:
    amount = parent_summary.head(n)["sum"].sum()
    share = amount / total_obligations * 100
    print(f"Top {n}: ${amount:,.2f} ({share:.2f}%)")
```

```python
agency_summary = (
    analysis_df
    .groupby("awarding_sub_agency_name")["federal_action_obligation"]
    .agg(["count", "sum", "mean"])
    .sort_values("sum", ascending=False)
)

print(agency_summary.head(20))
```

```python
agency_summary["share"] = (
    agency_summary["sum"] / total_obligations * 100
)

print(
    agency_summary[
        ["count", "sum", "share"]
    ].head(10)
)
```
